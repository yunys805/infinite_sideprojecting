```python
# <1. 엑셀윙스 연결>
import xlwings as xw

app = xw.App(add_book=False)
wb = app.books.open(r"C:\Users\tochd\OneDrive\바탕 화면\루틴 확인표(2026).xlsx")
ws = wb.sheets('sheet1')

# <2. 특정 행 밑 모든 표 전부 삭제>
# 1: 특정 행
ws.range('1:100000').delete()

# <3. 시작 표와 마지막 표 날짜 차이 계산>
from datetime import datetime

start = datetime(2026, 10, 12) # 시작 표(월) 날짜
end = datetime(2026, 12, 28) # 마지막 표(월) 날짜
gap = (end - start).days

print(gap)

# 결과:
#   2026.10.12(월)_시작 표(월) = 0일째
#   2026.12.28(월)_마지막 표(월) = 77일째
#   77일 차이

# <4. 생성해야 할 표 갯수 계산>
# 공식: 시작 표와의 날짜 차이 / 7
#   = '시작 표 이후' 몇 번째로 생성된 표인지
#   = 생성해야 할 표 갯수
#   설명: 시작 표 이후 1번째로 생성되는 표는 시작 표와 7일 차이가 난다.
#   그러므로, n번째로 생성된 표는 시작 표와 (7 * n)일 차이가 나야 한다.
#   그러므로, 공식은 (7 * n) / 7 = n이 되어야 하는 것이다.

# 예시1: 77일 차이 / 7 = 11
#   설명: 위 수식을 풀어 쓰면, (7 * 11)일 차이 / 7 = 11이 된다.
#   즉, 이것이 의미하는 바는,
#   시작 표와 마지막 표가 77(7 * 11)일 차이가 나면
#   시작 표 이후 11번째로 생성된 표가 마지막 표가 된다는 것이며,
#   그에 따라 내가 시작 표 이후 추가 생성해야 할 표의 갯수가 11개가 되어야 한다는 것이다.

# 예시2:
#   2026.10.12(월) = 시작 표
#   2026.10.19(월) = 1번째 표(7*1일 차이)
#   2026.10.26(월) = 2번째 표(7*2일 차이)
#   2026.11.02(월) = 3번째 표(7*3일 차이)
#   ...
#   2026.12.28(월) = 11번째 표(7*11일 차이)

# <5. 표 복제 생성>
# count = 몇 번째 표인지
# range('B740:J746'): 복사 대상인 시작 표의 범위
# 11: 생성해야 할 표의 갯수
# 740: 시작 표의 첫 행
# 9: 각 표의 첫 행과 첫 행 사이의 간격

ws.range('B740:J746').copy()

count = 1
while count <= 11:
    ws.range(f'B{740 + (9*count)}').paste()
    # count 1부터 (시작 표 이후) 1번째 표 생성
    # count 11은 (시작 표 이후) 11번째 표 생성
    count = count + 1

# <6. 생성해 놓은 표 날짜 변경>
# count2 = 몇 번째 표인지
# 11: 방금 생성해 놓은 표 갯수
# for문: 해당 행 D열-J열(날짜 줄)을 cell 변수로 하나씩 순회
# 740: 시작 표의 첫 행
# 9: 각 표의 첫 행과 첫 행 사이의 간격
# 원리:
#   1. 처음에 표를 생성해 놓으면, 모든 표의 날짜가 시작 표의 날짜와 전부 똑같게 된다.
#   2. 그래서 그 이후 아래 코드를 사용해, 그 시작 표의 날짜 + (7일 * n번째)로 덧셈하는 것.

from datetime import timedelta

count2 = 1
while count2 <= 11:
    for cell in ws.range(f'D{740 + 9*count2}:J{740 + 9*count2}'):
        # count2 1부터 (시작 표 이후) 1번째 표 날짜줄 변경
        # count2 11은 (시작 표 이후) 11번째 표 날짜줄 변경
        cell.value = cell.value + timedelta(days=(7*count2))
        # cell의 값(날짜 객체) + (7일 * n번째(날짜 단위)) 덧셈
    count2 = count2 + 1
```
